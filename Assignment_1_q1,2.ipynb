```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from google.colab import files
```

```python
uploaded = files.upload()
filename = list(uploaded.keys())[0]

im = np.array(Image.open(filename).convert('L'))
```

```python
def intensity_transform(im, breakpoints):
    breakpoints = np.array(breakpoints)

    x = breakpoints[:, 0]
    y = breakpoints[:, 1]

    # Create transformation for all possible intensity values
    input_levels = np.arange(256)

    # Piecewise-linear interpolation
    lookup_table = np.interp(input_levels, x, y)

    # Apply transformation
    transformed_im = lookup_table[im]

    return transformed_im.astype(np.uint8)
```

```python
breakpoints = np.array([
    [0,   0],
    [50,  50],
    [100, 150],
    [150, 255],
    [255, 255]
])

transformed_im = intensity_transform(im, breakpoints)
```

```python
plt.figure(figsize=(15, 5))


# Intensity transformation
plt.subplot(1, 3, 1)

plt.plot(
    breakpoints[:, 0],
    breakpoints[:, 1],
    marker='o'
)

plt.xlim(0, 255)
plt.ylim(0, 255)

plt.xlabel("Input intensity")
plt.ylabel("Output intensity")

plt.title("Intensity Transformation")
plt.grid()


# Original image
plt.subplot(1, 3, 2)

plt.imshow(im, cmap='gray', vmin=0, vmax=255)

plt.title("Original Image")
plt.axis('off')


# Transformed image
plt.subplot(1, 3, 3)

plt.imshow(transformed_im, cmap='gray', vmin=0, vmax=255)

plt.title("Transformed Image")
plt.axis('off')


plt.tight_layout()
plt.show()
```

```python
def intensity_transform(im, breakpoints):

    breakpoints = np.array(breakpoints)

    x = breakpoints[:, 0]
    y = breakpoints[:, 1]

    input_levels = np.arange(256)

    lookup_table = np.interp(input_levels, x, y)

    transformed_im = lookup_table[im]

    return transformed_im.astype(np.uint8)
```

White Matter

```python
white_breakpoints = np.array([
    [0,   0],
    [50,  20],
    [80,  50],
    [120, 190],
    [150, 220],
    [255, 255]
])

white_im = intensity_transform(im, white_breakpoints)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

plt.plot(
    white_breakpoints[:, 0],
    white_breakpoints[:, 1],
    marker='o'
)

plt.xlim(0, 255)
plt.ylim(0, 255)

plt.xlabel("Input intensity")
plt.ylabel("Output intensity")
plt.title("Transformation for White Matter")
plt.grid()


plt.subplot(1, 2, 2)

plt.imshow(white_im, cmap='gray', vmin=0, vmax=255)
plt.title("White Matter Accentuated")
plt.axis('off')

plt.tight_layout()
plt.show()
```

Gray Matter

```python
gray_breakpoints = np.array([
    [0,   0],
    [80,  30],
    [110, 60],
    [160, 210],
    [190, 235],
    [255, 255]
])

gray_im = intensity_transform(im, gray_breakpoints)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

plt.plot(
    gray_breakpoints[:, 0],
    gray_breakpoints[:, 1],
    marker='o'
)

plt.xlim(0, 255)
plt.ylim(0, 255)

plt.xlabel("Input intensity")
plt.ylabel("Output intensity")
plt.title("Transformation for Gray Matter")
plt.grid()


plt.subplot(1, 2, 2)

plt.imshow(gray_im, cmap='gray', vmin=0, vmax=255)
plt.title("Gray Matter Accentuated")
plt.axis('off')

plt.tight_layout()
plt.show()
```
